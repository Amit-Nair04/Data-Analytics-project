```python
# Converting CTC values to LPA (Ai Assisted)
def parse_ctc(x):
    if pd.isna(x):
        return x

    x = str(x).replace("₹", "").replace("LPA", "").replace(",", "").strip()

    value = float(x)

    if value > 100000:
        value = value / 100000

    return value
```

```python
# Converting  Current CTC and Previous CTC into numeric LPA
df["current_ctc"] = df["current_ctc"].apply(parse_ctc)
df["previous_ctc"] = df["previous_ctc"].apply(parse_ctc)
```

```python
# Calculating median, mean, minimum and maximum CTC for each role
role_ctc = df.groupby("role_clean")["current_ctc"].agg(
    Median="median",
    Mean="mean",
    Min="min",
    Max="max"
)

role_ctc = role_ctc.sort_values(by="Median", ascending=False)

role_ctc
```

### ANSWER Q3.1

#TASK 3 Q3.2

```python
# Creating experience bands for SDE Backend employees
sde_backend = df[df["role_clean"] == "SDE Backend"].copy()

bins = [0, 1, 3, 5, 100]
labels = ["0-1", "2-3", "4-5", "6+"]

sde_backend["exp_band"] = pd.cut(
    sde_backend["years_exp"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
# Calculating median CTC for each experience band (Ai Assisted)
exp_ctc = sde_backend.groupby("exp_band")["current_ctc"].median()

exp_ctc
```

```python
# Calculating percentage growth between experience bands
growth_23 = ((20.00 - 11.65) / 11.65) * 100
growth_45 = ((25.85 - 20.00) / 20.00) * 100
growth_6 = ((40.40 - 25.85) / 25.85) * 100

print(f"2-3 years: {growth_23:.2f}%")
print(f"4-5 years: {growth_45:.2f}%")
print(f"6+ years: {growth_6:.2f}%")
```

### ANSWER Q3.2

The median CTC for SDE Backend professionals increases from 11.65 LPA for 0–1 years of experience to 40.40 LPA for 6+ years. The largest salary jump (71.67%) occurs when moving from the 0–1 year band to the 2–3 year band, highlighting the value of gaining early industry experience.

```python
# Comparing median CTC of SDEs with and without System Design
with_sd = sde[sde["skills"].str.contains("System Design", na=False)]["current_ctc"].median()

without_sd = sde[~sde["skills"].str.contains("System Design", na=False)]["current_ctc"].median()

print("With System Design:", with_sd)
print("Without System Design:", without_sd)
```

```python
# Compare median CTC of SDEs with and without Kubernetes
with_k8 = sde[sde["skills"].str.contains("Kubernetes", na=False)]["current_ctc"].median()

without_k8 = sde[~sde["skills"].str.contains("Kubernetes", na=False)]["current_ctc"].median()

print("With Kubernetes:", with_k8)
print("Without Kubernetes:", without_k8)
```

# TASK 3 Q3.5

```python
# Creating experience bands for all employees (AI assisted)
bins = [0, 1, 3, 5, 100]
labels = ["0-1", "2-3", "4-5", "6+"]

df["exp_band"] = pd.cut(df["years_exp"], bins=bins, labels=labels, include_lowest=True)
```

```python
# Calculatingg group median CTC based on role, company type and experience
df["group_median"] = df.groupby(
    ["role_clean", "company_type_clean", "exp_band"]
)["current_ctc"].transform("median")
```

```python
# Counting employees in each group
df["group_size"] = df.groupby(
    ["role_clean", "company_type_clean", "exp_band"]
)["employee_id"].transform("count")
```

```python
# Calculating salary gap from the group median
df["gap"] = df["current_ctc"] - df["group_median"]
```

```python
# Displaying top 10 underpaid professionals
underpaid = df[df["group_size"] >= 10].sort_values("gap").head(10)

underpaid[[
    "employee_id",
    "role_clean",
    "company_type_clean",
    "years_exp",
    "current_ctc",
    "group_median",
    "gap"
]]
```

### ANSWER Q3.5

The most underpaid employee in the dataset is BLR0925 SDE Backend, MNC, earning 7.90 LPA below the median salary for professionals with the same role, company type, and experience level. Such comparisons can help employees identify whether they should negotiate a salary increase or explore better-paying opportunities.
